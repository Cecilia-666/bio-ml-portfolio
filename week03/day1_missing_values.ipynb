## Research Question

## Biological Context

Missing values frequently occur in:

- RNA-seq expression matrix
- Metabolomics intensity matrix
- Proteomics datasets

Improper handling may influence statistical analysis and machine learning models.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Create simulated RNA-seq expression matrix

Rows represent genes.
Columns represent samples.

This mimics a real expression matrix.

```python
np.random.seed(42)
expression_data = np.random.rand(10,5)*100
expression_data
```

```python
genes = [
    "Gene_A",
    "Gene_B",
    "Gene_C",
    "Gene_D",
    "Gene_E",
    "Gene_F",
    "Gene_G",
    "Gene_H",
    "Gene_I",
    "Gene_J"
]
len(genes)

samples=[
    "Control_1",
    "Control_2",
    "Control_3",
    "Treatment_1",
    "Treatment_2"
]
len(samples)
```

```python
df_expression = pd.DataFrame(
    expression_data,
    columns = samples,
    index = genes)
df_expression.shape
df_expression.head()
df_expression.columns
df_expression.index
```

```python
# Exercise 1
genes = [
    'BRCA1',
    'TP53',
    'EGFR',
    'MYC',
    'KRAS',
    'AKT1',
    'MTOR',
    'PTEN'
]
print(len(genes))
samples = [
    'Drug_A_1',
    'Drug_A_2',
    'Control_1',
    'Control_2'
]
print(len(samples))

np.random.seed(40)
expression_data = np.random.rand(8,4)*50

df_expression = pd.DataFrame(
    expression_data.round(2),
    index = genes,
    columns = samples)
display(df_expression)
type(df_expression)
df_expression.shape
df_expression.index
df_expression.columns
```

```python
df_expression.iloc[2,1] = np.nan
df_expression.iloc[5,3] = np.nan
df_expression
```

```python
# Check the NaN

print(df_expression.isna().sum())

# Calculate missing value rate for each sample
miss_rate = df_expression.isna().mean()
print(f"\n{miss_rate}")

print(f"\n{df_expression.isna().sum(axis=1)}")
```

```python
# Visualize missing value pattern
plt.figure(figsize=(6,4))

sns.heatmap(df_expression.isna(),
            cmap="viridis",
            cbar=False)

plt.title("Missing Values")
plt.xlabel("Samples")
plt.ylabel("Genes")

plt.show()
```
