```python
def calculate_gc(sequence):
    sequence = sequence.lower()
    g_count = sequence.count("g")
    c_count = sequence.count("c")
    total = len(sequence)

    if total == 0:
        return 0
    gc_percent = (g_count + c_count) / total * 100
    return gc_percent

gene1 = "ATGCGCTATCGC"
result = calculate_gc(gene1)
print(f"the result of {gene1} is {result:.1f} %")

sequences = ["ATGCGC", "ATATAT", "GCGCGC", "atgcatgc"]
for seq in sequences:
    gc = calculate_gc(seq)
    print(f"the gc of {seq} is {gc:.1f} %")
```

```python
def calculate_gc(sequence):
    sequence = sequence.lower()
    g_count = sequence.count("g")
    c_count = sequence.count("c")
    total = len(sequence)

    if total == 0:
        return 0
    gc_percent = (g_count + c_count) / total * 100
    return gc_percent

with open("test_sequence.txt", "w", encoding="utf-8") as f:
    f.write("ATGCGATCG\n")
    f.write("GCTATCGATCG\n")
    f.write("ATCGATCGATCG\n")
print("finished writing to file")
with open("test_sequence.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
    for line in lines:
        seq = line.strip()
        gc = calculate_gc(seq)
        print(f"the gc of {seq} is {gc:.1f} %")
```
